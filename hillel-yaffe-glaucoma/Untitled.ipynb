```python
import os

def show_folder_structure(path="."):
    for root, dirs, files in os.walk(path):
        # Skip hidden folders
        dirs[:] = [d for d in dirs if not d.startswith('.')]
        level = root.replace(path, '').count(os.sep)
        indent = '│   ' * level + '├── '
        print(f'{indent}{os.path.basename(root)}/')
        subindent = '│   ' * (level + 1) + '├── '
        for file in files:
            print(f'{subindent}{file}')

# Tukar path ni kepada folder YOLO anda
show_folder_structure("./hillel-yaffe-glaucoma")
```
