```python
"""
=============================================================
  Eye Image Quality Score Regression - Training Script
  Model     : MobileNetV2 (~3.4M params)
  Loss      : Huber Loss
  Optimizer : AdamW + CosineAnnealingLR
  Strategy  : 2-Phase training (freeze → unfreeze)
  Split     : 80 / 10 / 10  (train / val / test)
=============================================================
"""

import os
import random
import numpy as np
import pandas as pd
from PIL import Image
from pathlib import Path

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
from torch.optim.lr_scheduler import CosineAnnealingLR

from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score

import matplotlib.pyplot as plt
from tqdm import tqdm
import warnings
warnings.filterwarnings("ignore")

# ─────────────────────────────────────────────
# 0.  REPRODUCIBILITY
# ─────────────────────────────────────────────
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

# ─────────────────────────────────────────────
# 1.  CONFIG  ← tukar path di sini
# ─────────────────────────────────────────────
CSV_PATH   = r"C:\Users\Arif\Desktop\Project_Eyes\dataset\Labels.csv"
IMG_DIR    = r"C:\Users\Arif\Desktop\Project_Eyes\dataset\Images"
SAVE_DIR   = r"C:\Users\Arif\Desktop\Project_Eyes\outputs"

IMG_SIZE   = 224
BATCH_SIZE = 16
NUM_EPOCHS_PHASE1 = 10    # freeze backbone
NUM_EPOCHS_PHASE2 = 50    # unfreeze all
LR_PHASE1  = 1e-3
LR_PHASE2  = 1e-4
WEIGHT_DECAY = 1e-4
PATIENCE   = 10           # early stopping

os.makedirs(SAVE_DIR, exist_ok=True)
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {DEVICE}")

# ─────────────────────────────────────────────
# 2.  LOAD & INSPECT CSV
# ─────────────────────────────────────────────
df = pd.read_csv(CSV_PATH)
print("\n── CSV columns ──")
print(df.columns.tolist())
print(df.head())

# Auto-detect column names (flexible)
img_col   = df.columns[0]   # Image Name
score_col = df.columns[3]   # Quality Score  (index 3 = col D)

print(f"\nImage column   : '{img_col}'")
print(f"Score column   : '{score_col}'")
print(f"Score range    : {df[score_col].min():.3f}  →  {df[score_col].max():.3f}")
print(f"Score mean±std : {df[score_col].mean():.3f} ± {df[score_col].std():.3f}")
print(f"Total samples  : {len(df)}")

# Normalize score to [0, 1] for stable training
SCORE_MIN = df[score_col].min()
SCORE_MAX = df[score_col].max()
df["score_norm"] = (df[score_col] - SCORE_MIN) / (SCORE_MAX - SCORE_MIN)

# ─────────────────────────────────────────────
# 3.  TRAIN / VAL / TEST SPLIT  (80/10/10)
# ─────────────────────────────────────────────
train_df, temp_df = train_test_split(df, test_size=0.20, random_state=SEED)
val_df,   test_df = train_test_split(temp_df, test_size=0.50, random_state=SEED)

print(f"\nSplit → Train: {len(train_df)} | Val: {len(val_df)} | Test: {len(test_df)}")

# ─────────────────────────────────────────────
# 4.  DATASET
# ─────────────────────────────────────────────
class EyeDataset(Dataset):
    def __init__(self, dataframe, img_dir, transform=None):
        self.df        = dataframe.reset_index(drop=True)
        self.img_dir   = img_dir
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row      = self.df.iloc[idx]
        img_path = os.path.join(self.img_dir, row[img_col])
        image    = Image.open(img_path).convert("RGB")
        if self.transform:
            image = self.transform(image)
        score = torch.tensor(row["score_norm"], dtype=torch.float32)
        return image, score


# Augmentation for TRAIN only
train_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomVerticalFlip(),
    transforms.RandomRotation(degrees=15),
    transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std =[0.229, 0.224, 0.225]),
])

val_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std =[0.229, 0.224, 0.225]),
])

train_ds = EyeDataset(train_df, IMG_DIR, train_transform)
val_ds   = EyeDataset(val_df,   IMG_DIR, val_transform)
test_ds  = EyeDataset(test_df,  IMG_DIR, val_transform)

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,  num_workers=0, pin_memory=True)
val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False, num_workers=0, pin_memory=True)
test_loader  = DataLoader(test_ds,  batch_size=BATCH_SIZE, shuffle=False, num_workers=0, pin_memory=True)

# ─────────────────────────────────────────────
# 5.  MODEL  (MobileNetV2 regression head)
# ─────────────────────────────────────────────
class MobileNetV2Regressor(nn.Module):
    def __init__(self, pretrained=True):
        super().__init__()
        weights   = models.MobileNet_V2_Weights.DEFAULT if pretrained else None
        backbone  = models.mobilenet_v2(weights=weights)
        self.features = backbone.features          # output: (B, 1280, 7, 7)
        self.pool     = nn.AdaptiveAvgPool2d(1)    # → (B, 1280, 1, 1)
        self.head     = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(0.3),
            nn.Linear(1280, 256),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(256, 1),
            nn.Sigmoid()                           # output in [0,1] (normalised score)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.pool(x)
        x = self.head(x)
        return x.squeeze(1)                        # (B,)

    def freeze_backbone(self):
        for p in self.features.parameters():
            p.requires_grad = False

    def unfreeze_backbone(self):
        for p in self.features.parameters():
            p.requires_grad = True


model = MobileNetV2Regressor(pretrained=True).to(DEVICE)
total_params = sum(p.numel() for p in model.parameters())
print(f"\nTotal parameters : {total_params:,}  (~{total_params/1e6:.1f}M)")

# ─────────────────────────────────────────────
# 6.  LOSS & HELPERS
# ─────────────────────────────────────────────
criterion = nn.HuberLoss(delta=0.1)   # robust to outliers

def compute_metrics(preds_norm, targets_norm):
    """Return MAE, RMSE, R² in original score scale."""
    p = preds_norm   * (SCORE_MAX - SCORE_MIN) + SCORE_MIN
    t = targets_norm * (SCORE_MAX - SCORE_MIN) + SCORE_MIN
    mae  = np.mean(np.abs(p - t))
    rmse = np.sqrt(np.mean((p - t) ** 2))
    r2   = r2_score(t, p)
    return mae, rmse, r2

# ─────────────────────────────────────────────
# 7.  TRAINING LOOP
# ─────────────────────────────────────────────
def run_epoch(loader, train=True, optimizer=None, epoch=None, num_epochs=None, phase=None):
    model.train() if train else model.eval()
    total_loss, all_preds, all_targets = 0.0, [], []

    ctx = torch.enable_grad() if train else torch.no_grad()
    with ctx:
        for imgs, scores in loader:
            imgs, scores = imgs.to(DEVICE), scores.to(DEVICE)
            preds = model(imgs)
            loss  = criterion(preds, scores)
            if train:
                optimizer.zero_grad()
                loss.backward()
                optimizer.step()
            total_loss   += loss.item() * len(imgs)
            all_preds    += preds.detach().cpu().tolist()
            all_targets  += scores.cpu().tolist()

    avg_loss = total_loss / len(loader.dataset)
    mae, rmse, r2 = compute_metrics(np.array(all_preds), np.array(all_targets))
    return avg_loss, mae, rmse, r2


def train_phase(num_epochs, lr, phase_name, phase_short):
    optimizer = torch.optim.AdamW(
        filter(lambda p: p.requires_grad, model.parameters()),
        lr=lr, weight_decay=WEIGHT_DECAY
    )
    scheduler   = CosineAnnealingLR(optimizer, T_max=num_epochs, eta_min=lr * 0.01)
    best_val_loss = float("inf")
    patience_cnt  = 0
    history       = {"train_loss": [], "val_loss": [], "val_mae": [], "val_r2": []}

    print(f"\n{'='*65}")
    print(f"  {phase_name}")
    print(f"{'='*65}")

    epoch_bar = tqdm(range(1, num_epochs + 1), desc=f"  {phase_short}", unit="epoch",
                     bar_format="{l_bar}{bar:30}{r_bar}")

    for epoch in epoch_bar:
        tr_loss, tr_mae, tr_rmse, tr_r2 = run_epoch(train_loader, train=True,  optimizer=optimizer,
                                                      epoch=epoch, num_epochs=num_epochs, phase=phase_short)
        vl_loss, vl_mae, vl_rmse, vl_r2 = run_epoch(val_loader,   train=False,
                                                      epoch=epoch, num_epochs=num_epochs, phase=phase_short)
        scheduler.step()

        history["train_loss"].append(tr_loss)
        history["val_loss"].append(vl_loss)
        history["val_mae"].append(vl_mae)
        history["val_r2"].append(vl_r2)

        # Update outer epoch bar with summary metrics
        epoch_bar.set_postfix({
            "TrLoss": f"{tr_loss:.4f}",
            "VlLoss": f"{vl_loss:.4f}",
            "VlMAE" : f"{vl_mae:.4f}",
            "R²"    : f"{vl_r2:.4f}",
            "best"  : "✅" if vl_loss < best_val_loss else f"⏳{patience_cnt+1}/{PATIENCE}"
        })

        # Save best
        if vl_loss < best_val_loss:
            best_val_loss = vl_loss
            patience_cnt  = 0
            save_path = os.path.join(SAVE_DIR, "best_model.pth")
            os.makedirs(SAVE_DIR, exist_ok=True)   # pastikan folder wujud
            try:
                torch.save(model.state_dict(), save_path)
                tqdm.write(f"  💾 Model saved → {save_path}")
            except Exception as e:
                tqdm.write(f"  ❌ SAVE FAILED: {e}")
        else:
            patience_cnt += 1
            if patience_cnt >= PATIENCE:
                tqdm.write(f"  ⚠ Early stopping triggered at epoch {epoch}")
                break

    return history


# ── Phase 1: frozen backbone ──────────────────
model.freeze_backbone()
h1 = train_phase(NUM_EPOCHS_PHASE1, LR_PHASE1,
                 "PHASE 1 – Train head only (backbone frozen)", "P1")

# ── Phase 2: full fine-tune ───────────────────
model.unfreeze_backbone()
h2 = train_phase(NUM_EPOCHS_PHASE2, LR_PHASE2,
                 "PHASE 2 – Full fine-tune (all layers)", "P2")

# ── Save final weights (last epoch) ──────────
final_path = os.path.join(SAVE_DIR, "final_model.pth")
os.makedirs(SAVE_DIR, exist_ok=True)
torch.save(model.state_dict(), final_path)
print(f"\n💾 Final model saved → {final_path}")

# ─────────────────────────────────────────────
# 8.  TEST EVALUATION
# ─────────────────────────────────────────────
print("\n── Loading best model for test evaluation ──")
model.load_state_dict(torch.load(os.path.join(SAVE_DIR, "best_model.pth"), map_location=DEVICE))

ts_loss, ts_mae, ts_rmse, ts_r2 = run_epoch(test_loader, train=False, phase="TEST")
print(f"\n{'='*55}")
print(f"  TEST RESULTS")
print(f"  Loss (Huber) : {ts_loss:.4f}")
print(f"  MAE          : {ts_mae:.4f}")
print(f"  RMSE         : {ts_rmse:.4f}")
print(f"  R²           : {ts_r2:.4f}")
print(f"{'='*55}")

# ─────────────────────────────────────────────
# 9.  PLOTS
# ─────────────────────────────────────────────
# Combine phase histories
all_train_loss = h1["train_loss"] + h2["train_loss"]
all_val_loss   = h1["val_loss"]   + h2["val_loss"]
all_val_mae    = h1["val_mae"]    + h2["val_mae"]

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# (a) Loss curves
axes[0].plot(all_train_loss, label="Train Loss")
axes[0].plot(all_val_loss,   label="Val Loss")
axes[0].axvline(len(h1["train_loss"]) - 1, color="gray", linestyle="--", label="Phase 2 start")
axes[0].set_title("Huber Loss Curve")
axes[0].set_xlabel("Epoch"); axes[0].set_ylabel("Loss")
axes[0].legend(); axes[0].grid(True)

# (b) Val MAE
axes[1].plot(all_val_mae, color="orange", label="Val MAE")
axes[1].set_title("Validation MAE")
axes[1].set_xlabel("Epoch"); axes[1].set_ylabel("MAE (original scale)")
axes[1].legend(); axes[1].grid(True)

# (c) Prediction vs Actual (test set)
model.eval()
all_p, all_t = [], []
with torch.no_grad():
    for imgs, scores in test_loader:
        imgs = imgs.to(DEVICE)
        preds = model(imgs).cpu().numpy()
        all_p += list(preds   * (SCORE_MAX - SCORE_MIN) + SCORE_MIN)
        all_t += list(scores.numpy() * (SCORE_MAX - SCORE_MIN) + SCORE_MIN)

all_p, all_t = np.array(all_p), np.array(all_t)
axes[2].scatter(all_t, all_p, alpha=0.7, edgecolors="k", linewidths=0.5)
lims = [min(all_t.min(), all_p.min()) - 0.3,
        max(all_t.max(), all_p.max()) + 0.3]
axes[2].plot(lims, lims, "r--", label="Perfect prediction")
axes[2].set_xlim(lims); axes[2].set_ylim(lims)
axes[2].set_title(f"Predicted vs Actual  (R²={ts_r2:.3f})")
axes[2].set_xlabel("Actual Score"); axes[2].set_ylabel("Predicted Score")
axes[2].legend(); axes[2].grid(True)

plt.tight_layout()
fig.savefig(os.path.join(SAVE_DIR, "training_results.png"), dpi=150)
plt.show()
print(f"\n✅ Plot saved → {SAVE_DIR}\\training_results.png")
print(f"✅ Best model → {SAVE_DIR}\\best_model.pth")
```

```python
"""
=============================================================
  Eye Image Quality Score Regression - Inference Script
  - Predict semua gambar + label split (Train/Val/Test)
  - Evaluation metrics untuk Val & Test sahaja
  - Visualize 5 random gambar dari Test set
=============================================================
"""

import os
import random
import numpy as np
import pandas as pd
from PIL import Image
from pathlib import Path
from tqdm import tqdm

import torch
import torch.nn as nn
from torchvision import transforms, models
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
import matplotlib.pyplot as plt

# ─────────────────────────────────────────────
# CONFIG ← tukar path di sini
# ─────────────────────────────────────────────
MODEL_PATH  = r"C:\Users\Arif\Desktop\Project_Eyes\outputs\best_model.pth"
CSV_PATH    = r"C:\Users\Arif\Desktop\Project_Eyes\dataset\Labels.csv"
PREDICT_DIR = r"C:\Users\Arif\Desktop\Project_Eyes\dataset\Images"

# CSV_PATH   = r"C:\Users\Arif\Desktop\Project_Eyes\dataset\Labels.csv"
# IMG_DIR    = r"C:\Users\Arif\Desktop\Project_Eyes\dataset\Images"
# SAVE_DIR   = r"C:\Users\Arif\Desktop\Project_Eyes\outputs"


IMG_SIZE    = 224
SEED        = 42   # MESTI sama dengan train.py
DEVICE      = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {DEVICE}")

# ─────────────────────────────────────────────
# LOAD CSV + REPRODUCE SPLIT (sama macam train.py)
# ─────────────────────────────────────────────
df        = pd.read_csv(CSV_PATH)
img_col   = df.columns[0]
score_col = df.columns[3]
SCORE_MIN = df[score_col].min()
SCORE_MAX = df[score_col].max()
print(f"Score range: {SCORE_MIN:.3f} → {SCORE_MAX:.3f}")

# Reproduce exact same split dengan SEED yang sama
train_df, temp_df = train_test_split(df, test_size=0.20, random_state=SEED)
val_df,   test_df = train_test_split(temp_df, test_size=0.50, random_state=SEED)

# Buat lookup: filename → split label
split_lookup = {}
for fname in train_df[img_col]: split_lookup[fname] = "Train"
for fname in val_df[img_col]:   split_lookup[fname] = "Val"
for fname in test_df[img_col]:  split_lookup[fname] = "Test"

print(f"Split → Train: {len(train_df)} | Val: {len(val_df)} | Test: {len(test_df)}")

# Ground truth lookup
gt_lookup = dict(zip(df[img_col], df[score_col]))

# ─────────────────────────────────────────────
# MODEL
# ─────────────────────────────────────────────
class MobileNetV2Regressor(nn.Module):
    def __init__(self):
        super().__init__()
        backbone      = models.mobilenet_v2(weights=None)
        self.features = backbone.features
        self.pool     = nn.AdaptiveAvgPool2d(1)
        self.head     = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(0.3),
            nn.Linear(1280, 256),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(256, 1),
            nn.Sigmoid()
        )

    def forward(self, x):
        x = self.features(x)
        x = self.pool(x)
        x = self.head(x)
        return x.squeeze(1)


model = MobileNetV2Regressor().to(DEVICE)
model.load_state_dict(torch.load(MODEL_PATH, map_location=DEVICE))
model.eval()
print(f"✅ Model loaded from: {MODEL_PATH}")

# ─────────────────────────────────────────────
# TRANSFORM
# ─────────────────────────────────────────────
transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std =[0.229, 0.224, 0.225]),
])

# ─────────────────────────────────────────────
# PREDICT FUNCTION
# ─────────────────────────────────────────────
def predict_image(img_path: str) -> float:
    img    = Image.open(img_path).convert("RGB")
    tensor = transform(img).unsqueeze(0).to(DEVICE)
    with torch.no_grad():
        norm_score = model(tensor).item()
    return round(norm_score * (SCORE_MAX - SCORE_MIN) + SCORE_MIN, 4)

# ─────────────────────────────────────────────
# BATCH PREDICT SEMUA GAMBAR + LABEL SPLIT
# ─────────────────────────────────────────────
print("\n── Batch Prediction (All Images) ──")
results   = []
valid_ext = {".jpg", ".jpeg", ".png", ".bmp"}

for fname in tqdm(sorted(os.listdir(PREDICT_DIR)), desc="Predicting", unit="img"):
    if Path(fname).suffix.lower() not in valid_ext:
        continue
    full_path = os.path.join(PREDICT_DIR, fname)
    try:
        pred  = predict_image(full_path)
        gt    = gt_lookup.get(fname, None)
        split = split_lookup.get(fname, "Unknown")
        results.append({
            "Image"          : fname,
            "Split"          : split,
            "Ground_Truth"   : round(float(gt), 4) if gt is not None else None,
            "Predicted_Score": pred,
            "Abs_Error"      : round(abs(pred - float(gt)), 4) if gt is not None else None
        })
    except Exception as e:
        print(f"  ⚠ Skipped {fname}: {e}")

result_df = pd.DataFrame(results)

# Save full predictions CSV
out_csv = os.path.join(os.path.dirname(MODEL_PATH), "predictions_all.csv")
result_df.to_csv(out_csv, index=False)
print(f"\n✅ All predictions saved → {out_csv}")
print(result_df.to_string(index=False))

# ─────────────────────────────────────────────
# EVALUATION METRICS — VAL & TEST SAHAJA
# ─────────────────────────────────────────────
def eval_metrics(subset_df):
    gt   = subset_df["Ground_Truth"].values.astype(float)
    pred = subset_df["Predicted_Score"].values.astype(float)
    mae  = np.mean(np.abs(pred - gt))
    rmse = np.sqrt(np.mean((pred - gt) ** 2))
    r2   = r2_score(gt, pred)
    return mae, rmse, r2

val_df_r  = result_df[result_df["Split"] == "Val"]
test_df_r = result_df[result_df["Split"] == "Test"]

val_mae,  val_rmse,  val_r2  = eval_metrics(val_df_r)
test_mae, test_rmse, test_r2 = eval_metrics(test_df_r)

print(f"\n{'='*50}")
print(f"  EVALUATION RESULTS")
print(f"{'='*50}")
print(f"  {'Metric':<10} {'Val':>10} {'Test':>10}")
print(f"  {'-'*30}")
print(f"  {'MAE':<10} {val_mae:>10.4f} {test_mae:>10.4f}")
print(f"  {'RMSE':<10} {val_rmse:>10.4f} {test_rmse:>10.4f}")
print(f"  {'R²':<10} {val_r2:>10.4f} {test_r2:>10.4f}")
print(f"  {'N':<10} {len(val_df_r):>10} {len(test_df_r):>10}")
print(f"{'='*50}")

# ─────────────────────────────────────────────
# VISUALIZE 5 RANDOM DARI TEST SET
# ─────────────────────────────────────────────
print("\n── Visualize 5 Random Test Samples ──")

test_results = result_df[result_df["Split"] == "Test"].to_dict("records")
random.seed(SEED)
sample_5     = random.sample(test_results, min(5, len(test_results)))

fig, axes = plt.subplots(1, 5, figsize=(20, 5))
fig.suptitle(
    f"Test Set — Predicted vs Ground Truth  |  MAE={test_mae:.4f}  RMSE={test_rmse:.4f}  R²={test_r2:.4f}",
    fontsize=12, fontweight="bold"
)

for ax, item in zip(axes, sample_5):
    fname      = item["Image"]
    pred_score = item["Predicted_Score"]
    gt_score   = item["Ground_Truth"]
    error      = item["Abs_Error"]

    img = Image.open(os.path.join(PREDICT_DIR, fname)).convert("RGB")
    ax.imshow(img)
    ax.axis("off")

    color = "green" if error < 0.5 else "red"
    title = (f"{fname}\n"
             f"GT   : {gt_score:.4f}\n"
             f"Pred : {pred_score:.4f}\n"
             f"Err  : {error:.4f}")
    ax.set_title(title, fontsize=9, color=color)

plt.tight_layout()
out_fig = os.path.join(os.path.dirname(MODEL_PATH), "test_sample_predictions.png")
plt.savefig(out_fig, dpi=150, bbox_inches="tight")
plt.show()
print(f"✅ Visualization saved → {out_fig}")
```

```python
"""
=============================================================
  Eye Image Quality Score Regression - Training Script
  Model     : MobileNetV2 (~3.4M params) — architecture lama
  Loss      : MSE + Rank Loss (λ=0.3)
  Optimizer : AdamW + CosineAnnealingLR
  Strategy  : 2-Phase training (freeze → unfreeze)
  Split     : 80 / 10 / 10  (train / val / test)
=============================================================
"""

import os
import random
import numpy as np
import pandas as pd
from PIL import Image
from pathlib import Path

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
from torch.optim.lr_scheduler import CosineAnnealingLR

from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score

import matplotlib.pyplot as plt
from tqdm import tqdm
import warnings
warnings.filterwarnings("ignore")

# ─────────────────────────────────────────────
# 0.  REPRODUCIBILITY
# ─────────────────────────────────────────────
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

# ─────────────────────────────────────────────
# 1.  CONFIG
# ─────────────────────────────────────────────
CSV_PATH   = r"C:\Users\Arif\Desktop\Project_Eyes\dataset\Labels.csv"
IMG_DIR    = r"C:\Users\Arif\Desktop\Project_Eyes\dataset\Images"
SAVE_DIR   = r"C:\Users\Arif\Desktop\Project_Eyes\outputs"

IMG_SIZE          = 224
BATCH_SIZE        = 32
NUM_EPOCHS_PHASE1 = 15
NUM_EPOCHS_PHASE2 = 80
LR_PHASE1         = 3e-4
LR_PHASE2         = 5e-5
WEIGHT_DECAY      = 1e-3
PATIENCE          = 15
RANK_LAMBDA       = 0.3    # weight untuk Rank Loss

os.makedirs(SAVE_DIR, exist_ok=True)
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {DEVICE}")

# ─────────────────────────────────────────────
# 2.  LOAD CSV
# ─────────────────────────────────────────────
df = pd.read_csv(CSV_PATH)
print("\n── CSV columns ──")
print(df.columns.tolist())
print(df.head())

img_col   = df.columns[0]
score_col = df.columns[3]

print(f"\nImage column   : '{img_col}'")
print(f"Score column   : '{score_col}'")
print(f"Score range    : {df[score_col].min():.3f}  →  {df[score_col].max():.3f}")
print(f"Score mean±std : {df[score_col].mean():.3f} ± {df[score_col].std():.3f}")
print(f"Total samples  : {len(df)}")

SCORE_MIN = df[score_col].min()
SCORE_MAX = df[score_col].max()
df["score_norm"] = (df[score_col] - SCORE_MIN) / (SCORE_MAX - SCORE_MIN)

# ─────────────────────────────────────────────
# 3.  SPLIT  (80/10/10)
# ─────────────────────────────────────────────
train_df, temp_df = train_test_split(df, test_size=0.20, random_state=SEED)
val_df,   test_df = train_test_split(temp_df, test_size=0.50, random_state=SEED)
print(f"\nSplit → Train: {len(train_df)} | Val: {len(val_df)} | Test: {len(test_df)}")

# ─────────────────────────────────────────────
# 4.  DATASET
# ─────────────────────────────────────────────
class EyeDataset(Dataset):
    def __init__(self, dataframe, img_dir, transform=None):
        self.df        = dataframe.reset_index(drop=True)
        self.img_dir   = img_dir
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row      = self.df.iloc[idx]
        img_path = os.path.join(self.img_dir, row[img_col])
        image    = Image.open(img_path).convert("RGB")
        if self.transform:
            image = self.transform(image)
        score = torch.tensor(row["score_norm"], dtype=torch.float32)
        return image, score


train_transform = transforms.Compose([
    transforms.Resize((256, 256)),
    transforms.RandomCrop(224),
    transforms.RandomHorizontalFlip(),
    transforms.RandomVerticalFlip(),
    transforms.RandomRotation(degrees=20),
    transforms.ColorJitter(brightness=0.4, contrast=0.4, saturation=0.3, hue=0.1),
    transforms.RandomGrayscale(p=0.1),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std =[0.229, 0.224, 0.225]),
])

val_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std =[0.229, 0.224, 0.225]),
])

train_ds = EyeDataset(train_df, IMG_DIR, train_transform)
val_ds   = EyeDataset(val_df,   IMG_DIR, val_transform)
test_ds  = EyeDataset(test_df,  IMG_DIR, val_transform)

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,
                          num_workers=0, pin_memory=True)
val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False,
                          num_workers=0, pin_memory=True)
test_loader  = DataLoader(test_ds,  batch_size=BATCH_SIZE, shuffle=False,
                          num_workers=0, pin_memory=True)

# ─────────────────────────────────────────────
# 5.  MODEL — ARCHITECTURE LAMA (UNCHANGED)
# ─────────────────────────────────────────────
class MobileNetV2Regressor(nn.Module):
    def __init__(self, pretrained=True):
        super().__init__()
        weights       = models.MobileNet_V2_Weights.DEFAULT if pretrained else None
        backbone      = models.mobilenet_v2(weights=weights)
        self.features = backbone.features
        self.pool     = nn.AdaptiveAvgPool2d(1)
        self.head     = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(0.3),
            nn.Linear(1280, 256),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(256, 1),
            nn.Sigmoid()
        )

    def forward(self, x):
        x = self.features(x)
        x = self.pool(x)
        return self.head(x).squeeze(1)

    def freeze_backbone(self):
        for p in self.features.parameters():
            p.requires_grad = False

    def unfreeze_backbone(self):
        for p in self.features.parameters():
            p.requires_grad = True


model = MobileNetV2Regressor(pretrained=True).to(DEVICE)
total_params = sum(p.numel() for p in model.parameters())
print(f"\nTotal parameters : {total_params:,}  (~{total_params/1e6:.1f}M)")

# ─────────────────────────────────────────────
# 6.  LOSS — MSE + RANK LOSS
# ─────────────────────────────────────────────
mse_loss = nn.MSELoss()

def rank_loss(preds, targets):
    """
    Paksa model belajar relative ordering.
    Kalau gt_i > gt_j → pred_i patut > pred_j.
    """
    diff_pred   = preds.unsqueeze(1)   - preds.unsqueeze(0)    # (B, B)
    diff_target = targets.unsqueeze(1) - targets.unsqueeze(0)  # (B, B)
    sign        = torch.sign(diff_target)
    loss        = torch.clamp(1.0 - sign * diff_pred, min=0.0)
    # Upper triangle sahaja — elak duplicate pairs
    mask        = torch.triu(torch.ones_like(loss), diagonal=1).bool()
    return loss[mask].mean()

def combined_loss(preds, targets):
    return mse_loss(preds, targets) + RANK_LAMBDA * rank_loss(preds, targets)

def compute_metrics(preds_norm, targets_norm):
    p    = preds_norm   * (SCORE_MAX - SCORE_MIN) + SCORE_MIN
    t    = targets_norm * (SCORE_MAX - SCORE_MIN) + SCORE_MIN
    mae  = np.mean(np.abs(p - t))
    rmse = np.sqrt(np.mean((p - t) ** 2))
    r2   = r2_score(t, p)
    return mae, rmse, r2

# ─────────────────────────────────────────────
# 7.  TRAINING LOOP
# ─────────────────────────────────────────────
def run_epoch(loader, train=True, optimizer=None):
    model.train() if train else model.eval()
    total_loss, all_preds, all_targets = 0.0, [], []

    ctx = torch.enable_grad() if train else torch.no_grad()
    with ctx:
        for imgs, scores in loader:
            imgs, scores = imgs.to(DEVICE), scores.to(DEVICE)
            preds = model(imgs)
            loss  = combined_loss(preds, scores)
            if train:
                optimizer.zero_grad()
                loss.backward()
                optimizer.step()
            total_loss  += loss.item() * len(imgs)
            all_preds   += preds.detach().cpu().tolist()
            all_targets += scores.cpu().tolist()

    avg_loss = total_loss / len(loader.dataset)
    mae, rmse, r2 = compute_metrics(np.array(all_preds), np.array(all_targets))
    return avg_loss, mae, rmse, r2


def train_phase(num_epochs, lr, phase_name, phase_short):
    optimizer     = torch.optim.AdamW(
        filter(lambda p: p.requires_grad, model.parameters()),
        lr=lr, weight_decay=WEIGHT_DECAY
    )
    scheduler     = CosineAnnealingLR(optimizer, T_max=num_epochs, eta_min=lr * 0.01)
    best_val_loss = float("inf")
    patience_cnt  = 0
    history       = {"train_loss": [], "val_loss": [], "val_mae": [], "val_r2": []}

    print(f"\n{'='*65}")
    print(f"  {phase_name}")
    print(f"{'='*65}")

    epoch_bar = tqdm(range(1, num_epochs + 1), desc=f"  {phase_short}", unit="epoch",
                     bar_format="{l_bar}{bar:30}{r_bar}")

    for epoch in epoch_bar:
        tr_loss, tr_mae, _, tr_r2 = run_epoch(train_loader, train=True,  optimizer=optimizer)
        vl_loss, vl_mae, _, vl_r2 = run_epoch(val_loader,   train=False)
        scheduler.step()

        history["train_loss"].append(tr_loss)
        history["val_loss"].append(vl_loss)
        history["val_mae"].append(vl_mae)
        history["val_r2"].append(vl_r2)

        epoch_bar.set_postfix({
            "TrLoss": f"{tr_loss:.4f}",
            "VlLoss": f"{vl_loss:.4f}",
            "VlMAE" : f"{vl_mae:.4f}",
            "R²"    : f"{vl_r2:.4f}",
            "best"  : "✅" if vl_loss < best_val_loss else f"⏳{patience_cnt+1}/{PATIENCE}"
        })

        if vl_loss < best_val_loss:
            best_val_loss = vl_loss
            patience_cnt  = 0
            save_path     = os.path.join(SAVE_DIR, "best_model.pth")
            try:
                torch.save(model.state_dict(), save_path)
                tqdm.write(f"  💾 Model saved → {save_path}")
            except Exception as e:
                tqdm.write(f"  ❌ SAVE FAILED: {e}")
        else:
            patience_cnt += 1
            if patience_cnt >= PATIENCE:
                tqdm.write(f"  ⚠ Early stopping at epoch {epoch}")
                break

    return history


# ── Phase 1: frozen backbone ──────────────────
model.freeze_backbone()
h1 = train_phase(NUM_EPOCHS_PHASE1, LR_PHASE1,
                 "PHASE 1 – Train head only (backbone frozen)", "P1")

# ── Phase 2: full fine-tune ───────────────────
model.unfreeze_backbone()
h2 = train_phase(NUM_EPOCHS_PHASE2, LR_PHASE2,
                 "PHASE 2 – Full fine-tune (all layers)", "P2")

# ── Save final ───────────────────────────────
final_path = os.path.join(SAVE_DIR, "final_model.pth")
torch.save(model.state_dict(), final_path)
print(f"\n💾 Final model saved → {final_path}")

# ─────────────────────────────────────────────
# 8.  TEST EVALUATION
# ─────────────────────────────────────────────
print("\n── Loading best model for test evaluation ──")
model.load_state_dict(torch.load(
    os.path.join(SAVE_DIR, "best_model.pth"),
    map_location=DEVICE,
    weights_only=True
))

ts_loss, ts_mae, ts_rmse, ts_r2 = run_epoch(test_loader, train=False)
print(f"\n{'='*55}")
print(f"  TEST RESULTS  (MSE + Rank Loss)")
print(f"  Loss  : {ts_loss:.4f}")
print(f"  MAE   : {ts_mae:.4f}")
print(f"  RMSE  : {ts_rmse:.4f}")
print(f"  R²    : {ts_r2:.4f}")
print(f"{'='*55}")

# ─────────────────────────────────────────────
# 9.  PLOTS
# ─────────────────────────────────────────────
all_train_loss = h1["train_loss"] + h2["train_loss"]
all_val_loss   = h1["val_loss"]   + h2["val_loss"]
all_val_mae    = h1["val_mae"]    + h2["val_mae"]

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

axes[0].plot(all_train_loss, label="Train Loss")
axes[0].plot(all_val_loss,   label="Val Loss")
axes[0].axvline(len(h1["train_loss"]) - 1, color="gray",
                linestyle="--", label="Phase 2 start")
axes[0].set_title("MSE + Rank Loss Curve")
axes[0].set_xlabel("Epoch"); axes[0].set_ylabel("Loss")
axes[0].legend(); axes[0].grid(True)

axes[1].plot(all_val_mae, color="orange", label="Val MAE")
axes[1].set_title("Validation MAE")
axes[1].set_xlabel("Epoch"); axes[1].set_ylabel("MAE (original scale)")
axes[1].legend(); axes[1].grid(True)

model.eval()
all_p, all_t = [], []
with torch.no_grad():
    for imgs, scores in test_loader:
        imgs  = imgs.to(DEVICE)
        preds = model(imgs).cpu().numpy()
        all_p += list(preds * (SCORE_MAX - SCORE_MIN) + SCORE_MIN)
        all_t += list(scores.numpy() * (SCORE_MAX - SCORE_MIN) + SCORE_MIN)

all_p, all_t = np.array(all_p), np.array(all_t)
axes[2].scatter(all_t, all_p, alpha=0.7, edgecolors="k", linewidths=0.5)
lims = [min(all_t.min(), all_p.min()) - 0.3,
        max(all_t.max(), all_p.max()) + 0.3]
axes[2].plot(lims, lims, "r--", label="Perfect prediction")
axes[2].set_xlim(lims); axes[2].set_ylim(lims)
axes[2].set_title(f"Predicted vs Actual  (R²={ts_r2:.3f})")
axes[2].set_xlabel("Actual Score"); axes[2].set_ylabel("Predicted Score")
axes[2].legend(); axes[2].grid(True)

plt.tight_layout()
fig.savefig(os.path.join(SAVE_DIR, "training_results.png"), dpi=150)
plt.show()
print(f"\n✅ Plot saved → {SAVE_DIR}\\training_results.png")
print(f"✅ Best model → {SAVE_DIR}\\best_model.pth")
```
